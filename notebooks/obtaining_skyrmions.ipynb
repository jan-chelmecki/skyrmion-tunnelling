```julia
skyrmion_new = []
stable_list = []
for i = 100:50:500
    l0 = i / 100
    println("l0 = ", l0, "\n\n\n\n\n")

    filename = "../data/k01b12/l"*string(i)*".npy"
    n = npzread(filename)

    system = microscopic_system(length_scale=l0, k=0.1, b=1.2, n_sites=n.size[2])
    @unpack_system(system)

    centre_skyrmion!(n, system)
    symmetrise!(n,lattice)
    ST.perturb!(n,amp=0.3)
    relax!(n, system, N_steps=1000, graph=false)
    symmetrise!(n,lattice)
    stable = relax!(n, system, N_steps=10000, graph=true)

    push!(stable_list, stable)


    push!(skyrmion_new, n)
    println("l0 = ", l0)
    show_nz_heatmap(n,system.lattice)

    filename = "../data/k01b12/l"*string(i)*".npy"
    npzwrite(filename, n)
end
```
