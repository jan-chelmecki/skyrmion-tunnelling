```julia
using OrdinaryDiffEq
using SciMLBase


"""
    solve_ivp(u0, endpoint, system, coord; kwargs...)

Integrate the instanton equations adaptively from `u0` towards `endpoint`.

Returns
-------
t   : Vector{Float64}
    Adaptive time points.
sol : Matrix{ComplexF64}
    2 x length(t) matrix; each column is the state at the corresponding time.

The integration terminates when

    norm(u - endpoint) < endpoint_tol

or when the maximum integration time `T_max` is reached.

Use `direction = +1` for forward integration and `direction = -1`
for backward integration.
"""
function solve_ivp_new(u0::AbstractVector{<:Complex}, endpoint::AbstractVector{<:Complex}, system::System, coord::CollectiveCoordinate;
    direction::Int = 1, T_max::Real = 100.0,
    endpoint_tol::Real = 1e-8, reltol::Real = 1e-9, abstol::Real = 1e-11, dt_max::Real = Inf,
)

    @assert direction == 1 || direction == -1
    @assert T_max > 0

    @unpack_system(system)

    # Work arrays reused on every RHS evaluation.
    w  = zeros(ComplexF64, nx, ny)
    z  = zeros(ComplexF64, nx, ny)
    dw = zeros(ComplexF64, nx, ny)
    dz = zeros(ComplexF64, nx, ny)

    """
    OrdinaryDiffEq assumes the format:
    
         du/dt = rhs!(u, p, t)

    ----> define helper functions needed to communicate with the package
    """

    u0 = ComplexF64.(u0)
    endpoint = ComplexF64.(endpoint)

    p = (params = params, lattice = lattice, boundary = boundary, coord = coord, 
        w = w, z = z, dw = dw, dz = dz)

    # helper function for RHS
    function rhs!(du, u, p, t)
        v!(du, u[1], u[2], 
            p.w, p.z, p.dw, p.dz,
                p.params, p.lattice, p.boundary, p.coord
        )
        return nothing
    end

    function condition(u, t, integrator)
        return sqrt( abs2(u[1] - endpoint[1]) + abs2(u[2] - endpoint[2]) ) - endpoint_tol
    end
    # ContinuousCallback uses root finding, so the solver does not have to land exactly on the endpoint at the end of a timestep.

    function affect!(integrator)
        SciMLBase.terminate!(integrator)
    end

    callback = ContinuousCallback(condition, affect!,affect! )

    t_span = (0.0, direction * Float64(T_max))

    """
    Call the package solver
    """
    ODE_problem = ODEProblem(rhs!, u0, t_span, p)
    
    solve_kwargs = (reltol = reltol, abstol = abstol, callback = callback, save_everystep = true, save_start = true, save_end = true)

    # Optional maximum internal timestep.
    if isfinite(dt_max)
        solve_kwargs = merge(
            solve_kwargs,
            (dt_max = Float64(dt_max),)
        )
    end

    ODE_solution = solve(ODE_problem, Tsit5(); solve_kwargs...)

    # Extract time grid and solution matrix.
    t = collect(ODE_solution.t)
    sol = Array(ODE_Solution)

    # Check whether the endpoint was actually reached.
    final_distance = norm(sol[:, end] - endpoint)

    if final_distance > endpoint_tol
        @warn """
        Integration terminated at T_max before reaching the endpoint.
        Final distance = $final_distance
        Endpoint tolerance = $endpoint_tol
        """
    end

    return t, sol
end
```

```julia
function action(system::System, coord::CollectiveCoordinate, t::AbstractVector, sol::AbstractMatrix; show::Bool = true,)

    @assert size(sol, 1) == 2
    @assert size(sol, 2) == length(t)

    N = length(t) - 1

    @assert N >= 1

    # Adaptive timestep on every interval.
    Delta_t = diff(t)

    # The action density on each interval.
    # L[n] is associated with the midpoint of [t[n], t[n+1]].
    L = zeros(ComplexF64, N)

    @unpack_lattice system

    @inbounds for n in 1:N

        # Actual timestep on this interval.
        h = Delta_t[n]

        # State increments.
        Delta_x = sol[1, n+1] - sol[1, n]
        Delta_y = sol[2, n+1] - sol[2, n]

        # Midpoint state.
        x_mid = (sol[1, n] + sol[1, n+1]) / 2
        y_mid = (sol[2, n] + sol[2, n+1]) / 2

        # Accumulate the line integral over this interval.
        interval_action = 0.0 + 0.0im

        for j = 1:ny, i = 1:nx

            w, z, dw, dz =
                compute_wz_fields(coord, i, j, x_mid, y_mid)

            interval_action += ( z*dw*Delta_x - w*dz*Delta_y ) / (1 + w*z)

        end

        # Convert interval contribution to action density.
        L[n] = interval_action / h
    end

    # The action is the sum of the interval contributions.
    #
    # This is equivalent to:
    #
    #     sum(L[n] * Delta_t[n])
    #
    # but using interval_action directly is numerically a little cleaner.
    A = sum(L .* Delta_t)

    if show
        # Action density lives at the interval midpoints
        t_mid = (t[1:end-1] .+ t[2:end]) ./ 2

        P = plot(t_mid, real.(L), label = "real(L)", xlabel = L"t", ylabel = L"L", title = "Action density for the instanton")

        plot!(P, t_mid, imag.(L), label = "imag(L)")

        display(P)
    end

    println("\nTotal action = ", A)

    return A
end
```

```julia
function instanton_new(system::System, coord::CollectiveCoordinate; endpoint_forward, endpoint_backward, 
    T_forward_max::Real = 1e2, T_backward_max::Real = 1e2, endpoint_tol::Real = 1e-8,
    reltol::Real = 1e-9, abstol::Real = 1e-11, dt_max::Real = Inf,
    x_init = 1.0, direction_guess, dt=0.01, show_action=false, xlims=(-2,2))


    x_min = find_saddle(system,  coord, initial_guess = x_init)


    #x_min = 1.0
    H_inst = H(x_min,conj(x_min),system,coord)
    println("The saddle point is at ", x_min)

    x0, y0 = find_similar_energy(x_min, conj(x_min), system, coord, x_init = x_min+direction_guess[1], y_init=x_min+direction_guess[2])
    H0 = H(x0, y0, system, coord)
    println("The instantonic trajectory passes through the point \n x = ", x0, "\n y = ", y0)
    println("sanity check : |H_inst-H0| = ", abs(H_inst-H0))

    # The trajectory parting from thus found point is the instanton. Integrate it forward and backward in time.
    u0 = [x0,y0]

    # Forward integration towards endpoint_forward
    println("-- Integrating forwards --")

    t_forward, sol_forward = solve_ivp(u0, endpoint_forward, system, coord; 
                                    direction = +1, T_max = T_forward_max, endpoint_tol = endpoint_tol, 
                                    reltol = reltol, abstol = abstol)


    # Backward integration towards endpoint_backward
    println("-- Integrating backwards --")

    t_backward, sol_backward = solve_ivp(u0, endpoint_backward, system, coord;
                                        direction = -1, T_max = T_backward_max, endpoint_tol = endpoint_tol, 
                                        reltol = reltol, abstol = abstol)

    # Glue the solutions together
    t = vcat(reverse(t_backward),t_forward[2:end])
    sol = hcat(sol_backward[:, end:-1:1],sol_forward[:, 2:end])


    # Check that the time grid is strictly increasing.
    @assert all(diff(t) .> 0)


    println(
        "Forward endpoint distance = ",
        norm(sol_forward[:, end] - endpoint_forward)
    )

    println(
        "Backward endpoint distance = ",
        norm(sol_backward[:, end] - endpoint_backward)
    )

    S_inst = action(system, coord, t, sol; show = show_action)
    return S_inst, t, sol
end
```

```julia
# Forward integration towards endpoint_forward
println("-- Integrating forwards --")

t_forward, sol_forward = solve_ivp(u0, endpoint_forward, system, coord; 
                                direction = +1, T_max = T_forward_max, endpoint_tol = endpoint_tol, 
                                reltol = reltol, abstol = abstol)


# Backward integration towards endpoint_backward
println("-- Integrating backwards --")

t_backward, sol_backward = solve_ivp(u0, endpoint_backward, system, coord;
                                    direction = -1, T_max = T_backward_max, endpoint_tol = endpoint_tol, 
                                    reltol = reltol, abstol = abstol)


# ------------------------------------------------------------------
# Glue the two trajectories.
#
# t_backward runs
#
#     0, -dt1, -dt2, ..., -T
#
# so reverse it to obtain
#
#     -T, ..., -dt2, -dt1, 0.
#
# We omit the first point of the forward trajectory to avoid
# duplicating u0.
# ------------------------------------------------------------------

t = vcat(reverse(t_backward),t_forward[2:end])

sol = hcat(sol_backward[:, end:-1:1],sol_forward[:, 2:end])


# Check that the time grid is strictly increasing.
@assert all(diff(t) .> 0)


println(
    "Forward endpoint distance = ",
    norm(sol_forward[:, end] - endpoint_forward)
)

println(
    "Backward endpoint distance = ",
    norm(sol_backward[:, end] - endpoint_backward)
)


# ------------------------------------------------------------------
# Action
# ------------------------------------------------------------------

S_inst = action(system, coord, t, sol; show = show_action)
```
